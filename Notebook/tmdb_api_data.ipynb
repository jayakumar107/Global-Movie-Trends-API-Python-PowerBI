```python
url = "https://api.themoviedb.org/3/genre/movie/list?language=en"

response = requests.get(url, headers=headers)

genres = response.json()["genres"]

genre_df = pd.DataFrame(genres)

genre_df
```

```python
details = []

for movie_id in movies_df["Movie ID"]:

    url = f"https://api.themoviedb.org/3/movie/{movie_id}"

    response = requests.get(url, headers=headers)

    movie = response.json()

    details.append({
        "Movie ID": movie["id"],
        "Runtime": movie["runtime"],
        "Budget": movie["budget"],
        "Revenue": movie["revenue"],
        "Status": movie["status"],
        "Tagline": movie["tagline"],
        "Homepage": movie["homepage"],
        "IMDb": movie["imdb_id"],
        "Production Countries": ", ".join(
            [x["name"] for x in movie["production_countries"]]
        )
    })

details_df = pd.DataFrame(details)
```

```python
final_df = movies_df.merge(details_df, on="Movie ID")

final_df.head()
```

```python
#dropping duplicates of the same movies
inal_df = final_df.drop_duplicates(subset="Movie ID")
```

```python
# creating a new table called movie_genres
movie_genres = []

for _, row in movies_df.iterrows():
    movie_id = row["Movie ID"]

    for genre in row["Genre IDs"]:
        movie_genres.append({
            "Movie ID": movie_id,
            "Genre ID": genre
        })

movie_genres_df = pd.DataFrame(movie_genres)

movie_genres_df.head()
```
